2.  ¿Cuántos tickets se registraron en cada sucursal? ¿Cuál registró más?



```python
sucursalSur = dfVentas [dfVentas ['sucursal'] == "Sur"] ['id_ticket'].nunique ()
sucursalNorte = dfVentas[dfVentas ['sucursal'] == "Norte"] ['id_ticket'].nunique ()
sucursalCentro = dfVentas[dfVentas ['sucursal'] == "Central"] ['id_ticket'].nunique ()
print("En la sucursal Sur hay: ", sucursalSur, "tickets.")
print("En la sucursal Norte hay: ", sucursalNorte, "tickets.")
print("En la sucursal Centro hay: ", sucursalCentro, "tickets.")

totales = {
    'Sur': sucursalSur,
    'Norte': sucursalNorte, 'Central': sucursalCentro
}
# con key= totales.get, extrae el valor de las claves (los numeros), sin el, se extrae el nombre de los que esta dentro del diccionario (claves)
mayorSucursal = max (totales, key = totales.get)
# con totales[mayorSucursales] se va al diccionario totales y te da el valor de la que se haya calculado como mayorSucursal
print ("La sucursal con mas tickes es:", mayorSucursal, "con", totales [mayorSucursal], "tickets.")
```

6. ¿Cuál fue el importe total de ventas de cada sucursal? Calcule primero el precio total de cada línea como cantidad × precio_unitario.

```python
dfVentas ['importe_Ventas'] = dfVentas ['cantidad'] * dfVentas ['precio_unitario']

ventasPorSucursal = dfVentas.groupby ('sucursal') ['importe_Ventas'].sum ()
for sucursal, total in ventasPorSucursal.items():
    print(f"{sucursal}: ${total:,.2f}")
```

7.¿Cuántos tickets corresponden a compras de hasta 7 unidades, de 8 a 14 unidades y de 15 o más unidades? Sume primero las cantidades de todas las líneas de cada ticket.

```python
ticketsCantidades = dfVentas.groupby ('id_ticket') ['cantidad'].sum ()

hasta7 = 0
entre8y14=0
de15yMas= 0
for cantidad in ticketsCantidades:
    if cantidad <= 7:
        hasta7 += 1
    elif cantidad <=14:
        entre8y14 +=1
    else:
        de15yMas +=1

print ("Hasta 7 unidades: ", hasta7)
print ("Entre 8 y 14 unidades: ", entre8y14)
print ("Más de 15 unidades: ", de15yMas)
```

8.  ¿Cuántos tickets y qué importe total se registraron en cada mes? ¿En qué mes se alcanzó el mayor importe?



```python
dfVentas ['mes'] = dfVentas ['fecha'].astype (str).str [:7]

ticketsPorMes = dfVentas.groupby('mes') ['id_ticket'].nunique ()

importePorMes = dfVentas.groupby ('mes') ['importe_Ventas'].sum ()

print("========Tickets por mes========")
print (ticketsPorMes)
print("==========Importe por mes=========")
print (importePorMes)

print ("-----------------------")
print("El importe de ventas máximo fue en el mes de: ", importePorMes.idxmax())
```

9.¿Qué producto acumuló más unidades vendidas en cada sucursal? Indique su código, nombre y cantidad vendida. Si hay un empate, incluya todos los productos empatados.

```python
unirArchivos = pd.merge (dfVentas, dfProduct, on = 'id_producto')
resumen = unirArchivos.groupby (['sucursal', 'id_producto', 'producto']) ['cantidad'].sum ().reset_index()
# reset_index () para que se quite la primera columna de los indices y luego no me de error para filtrar resumen ['sucursal'] -> NO EXISTE
sucursales = resumen ['sucursal'].unique ()

for sucLes in sucursales:
    datosActuales = resumen [resumen ['sucursal']== sucLes]
    maxUnidades = datosActuales ['cantidad'].max ()
    ganador = datosActuales [datosActuales ['cantidad'] == maxUnidades]

    print (f"=========Sucursal: { sucLes } ====")
    for indice, fila in ganador.iterrows ():
        print (f"código: {fila ['id_producto']} | Poducto: {fila['producto']} | Unidades: {fila['cantidad']}")
```

10. Para cada tipo de cliente, ¿cuántos clientes distintos compraron, cuántos tickets generaron y cuál fue el importe total de sus compras?
